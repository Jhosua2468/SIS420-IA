```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms, models
import matplotlib.pyplot as plt
import os
import pandas as pd
import torch.nn.functional as F
```

```python
# Verificar si hay GPU disponible
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Usando dispositivo: {device}")
```

```python
# Rutas a tus carpetas
data_dir = 'C:/Users/saygu/IA/2-25-SIS420/Laboratorios/Lab 5/data/'
train_dir = os.path.join(data_dir, 'train')
test_dir = os.path.join(data_dir, 'test')
valid_dir = os.path.join(data_dir, 'valid')

# Verificar contenido de las carpetas
print("Archivos en data_dir:", os.listdir(data_dir))
print("Contenido de train_dir (primeros 5):", os.listdir(train_dir)[:5])

# Transformaciones para las imágenes
transform = transforms.Compose([
    transforms.Resize((224, 224)),  # Redimensionar
    transforms.ToTensor(),          # Convertir a tensor
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])  # Normalizar
])

# Cargar datasets con ImageFolder
train_dataset = datasets.ImageFolder(train_dir, transform=transform)
test_dataset = datasets.ImageFolder(test_dir, transform=transform)
valid_dataset = datasets.ImageFolder(valid_dir, transform=transform)
```

```python
# Crear DataLoaders
batch_size = 16  # Reducido para ahorrar memoria
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
valid_loader = DataLoader(valid_dataset, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

# Explorar
num_classes = len(train_dataset.classes)
print(f"Número de clases: {num_classes}")
print(f"Clases (primeras 10): {train_dataset.classes[:10]}...")
print(f"Imágenes en train: {len(train_dataset)}")
print(f"Imágenes en valid: {len(valid_dataset)}")
print(f"Imágenes en test: {len(test_dataset)}")

# Visualizar una imagen
#images, labels = next(iter(train_loader))
#plt.imshow(images[0].permute(1, 2, 0).clamp(0, 1))  # Corregir valores fuera de rango
#plt.title(f"Clase: {train_dataset.classes[labels[0]]}")
#plt.axis('off')
#plt.show()
```

```python
# 1. Modelo con Momentum (SGD con momentum)
class SimpleNet(nn.Module):
    def __init__(self, num_classes):
        super(SimpleNet, self).__init__()
        self.conv1 = nn.Conv2d(3, 32, kernel_size=3)
        self.pool = nn.MaxPool2d(2, 2)
        self.fc1 = nn.Linear(32 * 111 * 111, num_classes)  # Aproximado para 224x224 después de conv y pool

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))
        x = x.view(-1, 32 * 111 * 111)
        x = self.fc1(x)
        return x

# 2. Modelo con Batch Normalization
class NetWithBN(nn.Module):
    def __init__(self, num_classes):
        super(NetWithBN, self).__init__()
        self.conv1 = nn.Conv2d(3, 32, kernel_size=3)
        self.bn1 = nn.BatchNorm2d(32)
        self.pool = nn.MaxPool2d(2, 2)
        self.fc1 = nn.Linear(32 * 111 * 111, num_classes)

    def forward(self, x):
        x = self.pool(F.relu(self.bn1(self.conv1(x))))
        x = x.view(-1, 32 * 111 * 111)
        x = self.fc1(x)
        return x

# 3. Transfer Learning con ResNet18
model = models.resnet18(weights='IMAGENET1K_V1')
model.fc = nn.Linear(model.fc.in_features, num_classes)

# Congelar capas (transfer learning)
for param in model.parameters():
    param.requires_grad = False
for param in model.fc.parameters():
    param.requires_grad = True

model = model.to(device)

# Optimizador con momentum
optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=0.9)

# Scheduler para Learning Rate
scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.1)

# Función de pérdida
criterion = nn.CrossEntropyLoss()

# Entrenamiento
num_epochs = 10
for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    # Ajustar learning rate
    scheduler.step()

    # Validación
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for images, labels in valid_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {running_loss/len(train_loader):.4f}, Validation Accuracy: {100 * correct / total:.2f}%')

# Evaluación en test
model.eval()
correct = 0
total = 0
with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, predicted = torch.max(outputs.data, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

print(f'Test Accuracy: {100 * correct / total:.2f}%')

# Guardar el modelo
#torch.save(model.state_dict(), 'butterfly_model.pth')
#print("Modelo guardado como 'butterfly_model.pth'")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms, models
import matplotlib.pyplot as plt
import os
import pandas as pd

# Rutas a tus carpetas
data_dir = 'C:/Users/saygu/IA/2-25-SIS420/Laboratorios/Lab 5/data/'
train_dir = os.path.join(data_dir, 'train')
test_dir = os.path.join(data_dir, 'test')
valid_dir = os.path.join(data_dir, 'valid')

# Verificar contenido de las carpetas
print("Archivos en data_dir:", os.listdir(data_dir))
print("Contenido de train_dir:", os.listdir(train_dir)[:5])  # Muestra primeros 5 elementos

# Transformaciones para las imágenes
transform = transforms.Compose([
    transforms.Resize((224, 224)),  # Redimensionar a 224x224
    transforms.ToTensor(),          # Convertir a tensor
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225], inplace=True)  # Normalizar
])

# Cargar datasets con ImageFolder
try:
    train_dataset = datasets.ImageFolder(train_dir, transform=transform)
    test_dataset = datasets.ImageFolder(test_dir, transform=transform)
    valid_dataset = datasets.ImageFolder(valid_dir, transform=transform)
except Exception as e:
    print(f"Error al cargar los datasets: {e}")
    exit()

# Crear DataLoaders
batch_size = 16  # Reducido para ahorrar memoria
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
valid_loader = DataLoader(valid_dataset, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

# Explorar
num_classes = len(train_dataset.classes)
print(f"Número de clases: {num_classes}")
print(f"Clases: {train_dataset.classes[:10]}...")  # Muestra solo las primeras 10 para no saturar
print(f"Imágenes en train: {len(train_dataset)}")
print(f"Imágenes en valid: {len(valid_dataset)}")
print(f"Imágenes en test: {len(test_dataset)}")

# Intentar leer el CSV (opcional, para verificación)
training_csv = os.path.join(data_dir, 'training.csv.csv')
try:
    df_train = pd.read_csv(training_csv)
    print("Primeras filas de training.csv:")
    print(df_train.head())
except FileNotFoundError:
    print(f"Archivo {training_csv} no encontrado. Usando etiquetas de ImageFolder.")
except Exception as e:
    print(f"Error al leer el CSV: {e}")

# Visualizar una imagen
try:
    images, labels = next(iter(train_loader))
    plt.imshow(images[0].permute(1, 2, 0).clamp(0, 1))  # Clamp para corregir valores fuera de rango
    plt.title(f"Clase: {train_dataset.classes[labels[0]]}")
    plt.axis('off')
    plt.show()
except Exception as e:
    print(f"Error al visualizar imagen: {e}")

# Definir la red neuronal personalizada
class ButterflyNet(nn.Module):
    def __init__(self, num_classes):
        super(ButterflyNet, self).__init__()
        
        # Capa de entrada: Convolución inicial para procesar las imágenes 224x224x3
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(32)  # Batch Normalization después de la primera convolución
        
        # Capas ocultas: Capas convolucionales con pooling
        self.conv2 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, padding=1)
        self.bn2 = nn.BatchNorm2d(64)
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)
        
        self.conv3 = nn.Conv2d(in_channels=64, out_channels=128, kernel_size=3, padding=1)
        self.bn3 = nn.BatchNorm2d(128)
        
        # Calcular el tamaño de la imagen después de convoluciones y pooling
        # (224 -> 112 -> 56 -> 28 después de 3 poolings)
        self.fc1_input_size = 128 * 28 * 28  # Ajusta según el tamaño final
        
        # Capas totalmente conectadas (ocultas)
        self.fc1 = nn.Linear(self.fc1_input_size, 512)  # Primera capa oculta
        self.fc2 = nn.Linear(512, 256)                  # Segunda capa oculta
        self.dropout = nn.Dropout(0.5)                  # Dropout para regularización
        
        # Capa de salida
        self.fc3 = nn.Linear(256, num_classes)

    def forward(self, x):
        # Capa de entrada
        x = self.pool(torch.relu(self.bn1(self.conv1(x))))
        # Primera capa oculta
        x = self.pool(torch.relu(self.bn2(self.conv2(x))))
        # Segunda capa oculta
        x = self.pool(torch.relu(self.bn3(self.conv3(x))))
        
        # Aplanar para las capas totalmente conectadas
        x = x.view(x.size(0), -1)
        
        # Capas ocultas totalmente conectadas
        x = torch.relu(self.fc1(x))
        x = self.dropout(x)
        x = torch.relu(self.fc2(x))
        x = self.dropout(x)
        
        # Capa de salida
        x = self.fc3(x)
        return x

# Instanciar el modelo
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = ButterflyNet(num_classes=num_classes).to(device)

# Optimizar con Adam
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Scheduler para variar el learning rate
scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.1)

# Función de pérdida
criterion = nn.CrossEntropyLoss()

# Bucle de entrenamiento
num_epochs = 10
for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)
        
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item()
    
    # Actualizar learning rate
    scheduler.step()
    
    print(f"Epoch [{epoch+1}/{num_epochs}], Loss: {running_loss/len(train_loader):.4f}")
    
    # Validación
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for images, labels in valid_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    
    print(f"Validación Accuracy: {100 * correct / total:.2f}%")

# Evaluación en test
model.eval()
correct = 0
total = 0
with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, predicted = torch.max(outputs.data, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

print(f"Test Accuracy: {100 * correct / total:.2f}%")
```
