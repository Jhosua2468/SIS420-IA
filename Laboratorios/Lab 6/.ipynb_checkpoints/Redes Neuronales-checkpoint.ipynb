# 1. Importar librerías

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import torchvision.datasets as datasets
from torch.utils.data import DataLoader
import numpy as np
import matplotlib.pyplot as plt
```

# 2. Configuración GPU

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Dispositivo usado: {device}")
```

# 3. Cargar dataset SVHN

```python
# Configuración de ruta del dataset
svhn_root = "C:/Users/saygu/IA/Redes neuronales/SVHN"

# Transformaciones para normalizar imágenes
data_transforms = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

# Cargar dataset SVHN
train_dataset = datasets.SVHN(root=svhn_root, split='train', transform=data_transforms, download=False)
test_dataset = datasets.SVHN(root=svhn_root, split='test', transform=data_transforms, download=False)

# Configurar DataLoaders 
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True, num_workers=2)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False, num_workers=2)

# Verificar carga del dataset
print(f'Tamaño del dataset de entrenamiento: {len(train_dataset)}')
print(f'Tamaño del dataset de prueba: {len(test_dataset)}')
```

# 4. Arquitectura del Modelo

```python
# Definición de la red neuronal MLP
class MLP(nn.Module):
    def __init__(self):
        super(MLP, self).__init__()
        self.fc1 = nn.Linear(3 * 32 * 32, 1024)  # Aplanamos la imagen de 3x32x32
        self.fc2 = nn.Linear(1024, 512)  
        self.fc3 = nn.Linear(512, 256)  
        self.fc4 = nn.Linear(256,128)
        self.fc5 = nn.Linear(128,10)  # Capa de salida (10 clases)
        self.relu = nn.ReLU()

    def forward(self, x):
        # Aplanar imagen 3x32x32 → 3072 dimensiones
        x = x.view(x.size(0), -1)
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        x = torch.relu(self.fc3(x))
        x = torch.relu(self.fc4(x))
        # Salida (sin activación, usamos CrossEntropyLoss)
        x = self.fc5(x)
        return x

# Instanciar modelo y enviar a GPU
model = MLP().to(device)
print("\nArquitectura del modelo:")
print(model)
```

```python
print(torch.cuda.get_device_name(0))
```

# 5. Entrenamiento y Evaluacion

```python
# Definición de la función de pérdida y el optimizador
criterion = nn.CrossEntropyLoss()  # Función de pérdida para clasificación
optimizer = optim.Adam(model.parameters(), lr=0.0003) 

# Número de épocas 
num_epochs = 80  


# Listas para guardar métricas
train_losses = []
test_accuracies = []

# Bucle de entrenamiento
for epoch in range(num_epochs):
    model.train()  # Modo entrenamiento
    running_loss = 0.0
    
    for images, labels in train_loader:
        # Mover datos a GPU
        images, labels = images.to(device), labels.to(device)
        
        # Forward pass
        outputs = model(images)
        loss = criterion(outputs, labels)
        
        # Backpropagation
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item()
    
    # Calcular pérdida promedio por época
    avg_loss = running_loss / len(train_loader)
    print(f"Época [{epoch+1}/{num_epochs}], Pérdida: {avg_loss:.4f}")
    train_losses.append(avg_loss)

    # Evaluación del modelo en el conjunto de prueba
    model.eval()
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in test_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            _, predicted = torch.max(outputs, 1)  # Obtener la clase con mayor probabilidad
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    accuracy = 100 * correct / total
    test_accuracies.append(accuracy)
    print(f'Precisión del modelo en el conjunto de prueba: {accuracy:.2f}%')
```

# 6. Visualización de Resultados

```python
# Gráfica de pérdida y precisión
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(train_losses, label='Pérdida Entrenamiento')
plt.xlabel('Época')
plt.ylabel('Pérdida')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(test_accuracies, label='Precisión Test', color='orange')
plt.xlabel('Época')
plt.ylabel('Precisión (%)')
plt.legend()
plt.suptitle('Progreso del Entrenamiento')
plt.show()
```

# 7. Visualización de Capas Ocultas

```python
# Obtener un ejemplo del test set
dataiter = iter(test_loader)
images, labels = next(dataiter)
sample_image = images[0:1].to(device)  # Tomar 1 imagen

# Extraer activaciones de todas las capas ocultas
model.eval()
with torch.no_grad():
    out1 = torch.relu(model.fc1(sample_image.view(1, -1)))
    out2 = torch.relu(model.fc2(out1))
    out3 = torch.relu(model.fc3(out2))
    out4 = torch.relu(model.fc4(out3))

    # Convertir a numpy
    act1 = out1.cpu().numpy().flatten()
    act2 = out2.cpu().numpy().flatten()
    act3 = out3.cpu().numpy().flatten()
    act4 = out4.cpu().numpy().flatten()

# Graficar activaciones de todas las capas ocultas
plt.figure(figsize=(20, 10))

plt.subplot(2, 2, 1)
plt.bar(range(len(act1)), act1, alpha=0.7)
plt.title("Activaciones Capa Oculta 1 (512 neuronas)")
plt.xlabel("Índice de Neurona")
plt.ylabel("Valor de Activación")

plt.subplot(2, 2, 2)
plt.bar(range(len(act2)), act2, alpha=0.7, color='orange')
plt.title("Activaciones Capa Oculta 2 (256 neuronas)")
plt.xlabel("Índice de Neurona")

plt.subplot(2, 2, 3)
plt.bar(range(len(act3)), act3, alpha=0.7, color='green')
plt.title("Activaciones Capa Oculta 3 (128 neuronas)")
plt.xlabel("Índice de Neurona")

plt.subplot(2, 2, 4)
plt.bar(range(len(act4)), act4, alpha=0.7, color='red')
plt.title("Activaciones Capa Oculta 4 (64 neuronas)")
plt.xlabel("Índice de Neurona")

plt.tight_layout()
plt.show()

# Mostrar imagen de ejemplo
plt.imshow(images[0].permute(1, 2, 0).cpu().numpy() * 0.5 + 0.5)  # Desnormalizar
plt.title(f"Etiqueta Real: {labels[0].item()}")
plt.axis('off')
plt.show()
```
